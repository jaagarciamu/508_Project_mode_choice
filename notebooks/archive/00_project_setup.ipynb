# 00 — Configuración del proyecto

## Objetivo
Confirmar que el notebook se ejecuta desde una copia del repositorio y que la configuración básica está disponible, sin abrir ni procesar microdatos.

## Requisitos
Python 3.12 y dependencias sincronizadas mediante `uv sync` desde la raíz del proyecto.

```python
from pathlib import Path
import yaml

candidates = (Path.cwd(), *Path.cwd().parents)
PROJECT_ROOT = next((path for path in candidates if (path / "pyproject.toml").is_file()), None)
if PROJECT_ROOT is None:
    raise RuntimeError("No se encontró pyproject.toml en el directorio actual ni en sus directorios padre.")

with (PROJECT_ROOT / "configs" / "project.yaml").open(encoding="utf-8") as file:
    project_config = yaml.safe_load(file)

print(f"Project root: {PROJECT_ROOT}")
print(f"Project: {project_config['project']['name']}")
print(f"Current level: {project_config['project']['current_level']}")
print(f"Raw source directory exists: {(PROJECT_ROOT / project_config['data_governance']['raw_data_path']).is_dir()}")
```
